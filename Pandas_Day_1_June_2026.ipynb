# Filtering Dataframe

```python
df.loc[df["mark"]>90, ["name", "class", "mark"]]
```

```python
# df.iloc[df["mark"]>90, [1,2,3]]
```

```python
df.loc[df["class"]=="Six", :]
# : means return all range
```

```python
df.loc[df["gender"]=="female", ["name", "gender"]]
```

- filter by multiple condition

```python
df.loc[(df["class"]=="Five") & (df["gender"]=="male"), ["name", "class", "gender"]]
```

```python
df.loc[(df["class"]=="Seven") & (df["mark"]>=70), ["name", "class", "mark"]]
```

```python
# write condition insidei [] to filter dataframe using slicing method
df[df["mark"]>85]
```

```python
df[df["class"]=="Four"]
```

```python
df[(df["class"]=="Four") & (df["gender"]=="female")]
```

```python
df[(df["class"]=="Four") |(df["gender"]=="female")]
# filter by class four or gender female
```

```python
df.filter(items=['name', 'mark'])
```

```python
df.filter(like='me')
```

```python
df.query("mark > 60")
```

```python
df.query('gender == "female"')
```
